```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt, matplotlib.dates as mdates
from matplotlib.dates import DateFormatter

dates = pd.date_range(start= '2023-01-01', periods=100)
values = np.random.rand(100).cumsum()
data= pd.DataFrame(
    {
        'Date': dates,
        'Value': values
    })

data
```

```python
# Crear gráfico de líneas

fig, ax = plt.subplots(figsize=(12,6))

ax.plot(data['Date'], data['Value'], color='green')
ax.xaxis.set_major_locator(mdates.DayLocator(interval=5))
marcas_eje_y = np.arange(0, 60, 5)

plt.yticks(marcas_eje_y)
plt.xlabel('Date')
plt.ylabel('Value')
plt.title('Serie de tiempo')
plt.grid()
plt.xticks(rotation=45)
plt.tight_layout()
```

```python
from matplotlib.dates import DateFormatter

dates = pd.date_range(start= '2023-01-01', periods=12, freq='ME')
sales = np.random.randint(1000, 5000, 12)
yticks = np.arange(0, 5000, 200)

df = pd.DataFrame({
        'Date': dates,
        'Sales': sales
    })


plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%m/%Y'))
plt.gca().xaxis.set_major_locator(mdates.MonthLocator())
plt.yticks(yticks)
plt.plot(df['Date'], df['Sales'], marker='o', linestyle= '-', label = 'Ventas mensuales')
plt.title('Análisis de ventas mensuales')
plt.xticks(rotation=45)
plt.xlabel('Date')
plt.ylabel('Sales')
plt.legend()
plt.tight_layout()
plt.grid()
```
