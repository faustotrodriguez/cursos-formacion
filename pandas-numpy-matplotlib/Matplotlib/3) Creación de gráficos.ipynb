```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))

categorias = ['Producto A', 'Producto B', 'Producto C']
ventas = [120, 150, 90]

# Creación de gráfico de barras verticales
plt.bar(categorias, ventas, color= 'skyblue', label= 'Ventas mensuales')
plt.title('Ventas de productos en un mes')
plt.xlabel('Productos')
plt.ylabel('Ventas')
plt.legend(loc='best',fontsize= 11, frameon= False)
plt.annotate('Máximo de ventas', xy=(0.684, 140))
plt.annotate('Mínimo de ventas', xy=(1.684,96))
plt.annotate('Producto destacado',xy=(1,1), xytext=(-0.350,140), arrowprops= dict(facecolor='green', shrink=0.055))
plt.grid(True, alpha=0.3)
```

```python
productos = ['Producto A', 'Producto B', 'Producto C']
ventas_p = [50, 35, 15]

plt.pie(ventas_p, labels=productos, autopct='%1.1f%%', startangle= 140, colors=['gold', 'lightcoral', 'orange'])
plt.title('Participación de Mercado por producto', fontdict= {'fontweight':'bold', 'fontfamily':'sans-serif', 'fontstyle':'italic'})
plt.axis('equal')
```
