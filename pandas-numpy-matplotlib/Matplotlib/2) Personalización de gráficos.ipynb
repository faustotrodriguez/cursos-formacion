```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt

hours = [2, 3, 4, 5, 6, 7, 8]
exam_student_1 = [55, 60, 65, 70, 75, 80, 85]
exam_student_2 = [50, 58, 63, 69, 74, 78, 83]

plt.figure(figsize=(8,4))

# Creo el gráfico de dispersión para 2 estudiantes
plt.scatter(hours, exam_student_1, color='green', marker= 'o', linestyle= '-')
plt.scatter(hours, exam_student_2, color= 'red')
```

```python
# Otra manera de resolverlo por ejemplo con plot, dandole el estilo de las lineas

plt.plot(hours, exam_student_1, color='green', marker='o', linestyle='--', linewidth= 2, label='Estudiante 1')
plt.plot(hours, exam_student_2, color='red', marker='s', linestyle='--', linewidth= 0.5, label='Estudiante 2')
plt.legend(loc='lower right', fontsize= 12, frameon= False)
plt.title('Relación horas estudio/resutado entre dos estudiantes')
```
