```python
import numpy as np, pandas as pd, matplotlib.pyplot as plt

# Creo 2 arrays para hacer las primeras pruebas con matplotlib
month = np.array(['Enero','Febrero','Marzo','Abril','Mayo'])
sales = np.array([20, 25, 30, 28, 35])

# Configuro el tamaño del grafico
plt.figure(figsize=(8,6))

# Creo el grafico
plt.plot(month, sales, marker= 'o', color='red')

# Agrego un titulo
plt.title('Ventas mensuales')

# Agrego subtitulos en los ejes
plt.xlabel('Meses')
plt.ylabel('Cantidad de ventas')
```
