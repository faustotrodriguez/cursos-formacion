```python
import pandas as pd

df1 = pd.DataFrame({
  'key': ['A', 'B', 'C'],
  'value1': [1,2,3]
})

df2 = pd.DataFrame({
  'key': ['B', 'C', 'D'],
  'value2': [4,5,6]
})

print(df1,df2)
```

***Merge en pandas***

```python
# INTERSECCIÓN: toma solo lo que coincida entre el df1 y el df2 (inner) 

inner_merged = pd.merge(df1, df2, how= 'inner', on= 'key')
inner_merged
```

```python
# UNION: toma todo lo que haya en los dfs (outer)

outer_merged = pd.merge(df1, df2, how= 'outer', on= 'key')
outer_merged
```

```python
# Toma todo lo del df de la izquierda, y solo lo que coincida del df de la derecha

left_merged = pd.merge(df1, df2, how= 'left', on= 'key')
left_merged
```

```python
# Toma todo lo del df de la derecha, y solo lo que coincida del df de la izquierda

right_merged = pd.merge(df1, df2, how= 'right', on= 'key')
right_merged
```

***Concatenación en pandas***

```python
df3 = pd.DataFrame({
  'A': ['A0', 'A1', 'A2'],
  'B': ['B0', 'B1', 'B2'],
})

df4 = pd.DataFrame({
  'A': ['A3', 'A4', 'A5'],
  'B': ['B3', 'B4', 'B5'],
}) 

print(df3,df4)
```

```python
vertical_concat = pd.concat([df3, df4])
vertical_concat
```

```python
hotizontal_concat = pd.concat([df3, df4], axis=1)
hotizontal_concat
```

```python
df5 = pd.DataFrame({
    'A': ['A0', 'A1', 'A2'],
    'B': ['B0', 'B1', 'B2']
  }, 
  index = ['K0', 'K1', 'K2']
) 

df6 = pd.DataFrame({
    'C': ['C0', 'C1', 'C2'],
    'D': ['DO', 'D1', 'D2']
  }, 
  index = ['K0', 'K2', 'K3']
)

print(df5, df6)
```

```python
joined = df5.join(df6, how= 'inner')
joined
```
